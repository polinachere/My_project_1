# Создание нового признака

уровень опыта по категориям

Метод .apply() применяет функцию к каждому элементу Series

```python
def experience_level(exp):
  if exp < 3:
    return 'Junior'
  elif exp < 7:
    return 'Mid'
  elif exp < 15:
    return 'Senior'
  else:
    return 'Expert'

df['Experience Level'] = df['Years of Experience'].apply(experience_level)
print(df['Experience Level'].value_counts())
print (df)
```

зависимость з/п от стажа работы

max() возвращает максимальное из двух значений

Если опыт = 0 → max(0, 1) = 1

Если опыт = 5 → max(5, 1) = 5
Защита от деления на ноль

Если опыт = 0 лет, делим на 1, а не на 0

Предотвращает ошибку ZeroDivisionError
axis=1 - применяется к строкам

```python
df['Salary per year'] = df.apply(lambda row: row['Salary']/max(row['Years of Experience'], 1), axis = 1)
print(df.groupby('Experience Level')['Salary per year'].mean().round(2))
print(df)
```

# Группировка и агрегация

средняя з/п по странам и уровню образования

Метод для агрегации (суммирования) данных в группах

Применяет указанные функции к каждой группе

Для столбца 'Salary' вычисляет:

mean - среднее арифметическое

median - медиану (серединное значение)

count - количество элементов в группе

Для столбца 'Years of Experience' вычисляет:

mean - средний опыт работы

```python
groupirovka1 = df.groupby(['Country', 'Education Level']).agg({
    'Salary': ['mean', 'median', 'count'],
    'Years of Experience': 'mean'
}).round(2)

print(groupirovka1)
```

зависимость з/п от пола и уровня работы

std(), var() — стандартное отклонение и дисперсия

```python
groupirovka2 = df.groupby(['Experience Level', 'Gender']).agg({
    'Salary': ['mean', 'std', 'count'],
    'Age': 'mean'
}).round(2)

print(groupirovka2)
```

группировка должностей с расчетом средних показателей

```python
groupirovka3 = df.groupby('Job Title').agg({
    'Salary':['mean','count','max','min'],
    'Years of Experience': 'mean'
}).round(2)

print(groupirovka3)
```

# Визуализация

plt.style.use('default') - базовые настройки

далее указана ширина и высота графика

вычиляем среднюю з/п сортируем по убыванию!!

в графика х - названия стран, а у-з/п
plt.xticks - поворачиваем надписи на оси х для читаемости

добавляем числовые значения:
enumerate() - проходит по значениям вместе с их индексами

i - индекс столбца (0, 1, 2...)

value - числовое значение зарплаты

plt.text() - добавляет текст на график

i - X-координата текста (центр столбца)
value + 1000 - Y-координата текста (немного выше вершины столбца)

f'${value:,.0f}' - форматирование

 :, - разделитель тысяч (1,000 вместо 1000)
 .0f - 0 десятичных знаков (целое число)

 ha='center' - горизонтальное выравнивание по центру

fontsize=9 - размер шрифта

```python
plt.style.use('default')
plt.figure(figsize=(10,6))

country_salary = df.groupby('Country')['Salary'].mean().sort_values(ascending=False)

country_salary.plot(kind = 'bar', color='skyblue')
plt.title('Средняя зарплата по странам')
plt.xlabel('Страна')
plt.ylabel('Средняя з/п')
plt.xticks(rotation=45)

for i, value in enumerate(country_salary.values):
    plt.text(i, value + 1000, f'${value:,.0f}', ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
plt.style.use('default')
plt.figure(figsize=(12,6))

job_salary = df.groupby('Job Title')['Salary'].mean().sort_values(ascending=False)

job_salary.head(10).plot(kind= 'bar', color='lightcoral')
plt.title('Средняя зарплата по Должностям (топ-10)')
plt.xlabel('должность')
plt.ylabel('Средняя з/п')
plt.xticks(rotation=45)

for i, value in enumerate(job_salary.values[:10]):
    plt.text(i, value + 500, f'${value:,.0f}', ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

unstack() - это ключевая операция, которая преобразует мультииндекс
 в удобный для визуализации формат:
 - Страны станут строками (индекс)
- Уровни образования станут столбцами
- В ячейках будут средние зарплаты

kind='bar' - тип графика: столбчатая диаграмма

width=0.8 - ширина столбцов (0.8 = 80% от доступного пространства)

 Остальные 20% - промежутки между группами столбцов

figsize=(14, 7) - дублируем размер (можно было не указывать, так как уже задали в plt.figure)


bbox_to_anchor=(1.05, 1) - перемещает легенду ЗА пределы области графика:

1.05 по X = на 5% правее правой границы графика

1 по Y = на уровне верхней границы графика

loc='upper left' - якорная точка легенды (левый верхний угол легенды)


Это предотвращает наложение легенды на сами столбцы

```python
grouped = df.groupby(['Country', 'Education Level'])['Salary'].mean().unstack()

plt.figure(figsize=(14, 7))
grouped.plot(kind='bar', width=0.8, figsize=(14, 7))
plt.title('Средняя зарплата по странам и уровню образования', fontsize=16)
plt.xlabel('Страна')
plt.ylabel('Средняя зарплата ($)')
plt.legend(title='Уровень образования', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
